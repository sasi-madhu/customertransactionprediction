<h2>DATA PREPARATION</h2>


```python
#remove unneccesary features
df = df.drop(columns=['ID_code'])
df.head()
```

```python
#Set the columns for train,test
X = df.drop(columns='target')
y = df['target']
```

```python
y.head()
```

```python
#Model Building
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,      # 20% goes to testing
    random_state=42,    # For reproducibility
    stratify=y          # Keep class ratio same in train and test
)


print(X_train.shape)
print(X_test.shape)
```

<h2>Data Preprocessing</h2>

```python
#Scaling the features for better predictions like logistic regression algorithm
#Mean 0, standard deviation=1
from sklearn.preprocessing import StandardScaler
import warnings
warnings.filterwarnings('ignore')
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

<h1>Model Building</h1>
<h2>1. Logistic Regression</h2>

```python
from sklearn.linear_model import LogisticRegression
import warnings
warnings.filterwarnings('ignore')

lr_model = LogisticRegression(class_weight='balanced',  # Handles imbalanced data automatically
    max_iter=1000,            
    random_state=42)
lr_model.fit(X_train_scaled,y_train)
```

```python
#Prediction
lr_y_pred = lr_model.predict(X_test_scaled)
lr_y_prob =  lr_model.predict_proba(X_test_scaled)[:, 1]
```

```python
lr_y_pred
```

```python
#Evaluation
from sklearn.metrics import (
    accuracy_score, classification_report,
    confusion_matrix, roc_auc_score, roc_curve
)
lr_acc = accuracy_score(y_test,lr_y_pred)
lr_roc = roc_auc_score(y_test,lr_y_prob)
print(f"\nLogistic Regression Results:")
print(f"   Accuracy  : {lr_acc:.4f} ({lr_acc*100:.2f}%)")
lr_prob = lr_model.predict_proba(X_test_scaled)[:, 1]
print(f"   ROC-AUC   : {lr_roc:.4f}")
```

```python
#Classification report
print(classification_report(y_test,lr_y_pred,target_names=["No Trasanction","Will Transact"]))
print(confusion_matrix(y_test, lr_y_pred))
```

<h2>Random Forest</h2>


```python
#Model building

from sklearn.ensemble import RandomForestClassifier
import warnings
warnings.filterwarnings('ignore')


print("Random Forest result")
print("This may take 2-3 min....")

rf_model = RandomForestClassifier(
    n_estimators=100,         # Build 100 decision trees
    class_weight={0:1, 1:5},  # Handle imbalanced classes
    n_jobs=-1,                # Use all CPU cores (faster!)
    random_state=42
)


#Model training & Testing
rf_model.fit(X_train_scaled,y_train)


#rf_y_pred = rf_model.predict(X_test_scaled)
rf_y_prob = rf_model.predict_proba(X_test_scaled)[:, 1]
rf_y_pred = (rf_y_prob >= 0.2).astype(int)


#Evaluation
rf_acc = accuracy_score(y_test,rf_y_pred)
rf_roc = roc_auc_score(y_test,rf_y_prob)


print(f"Accuracy : {rf_acc:.4f}({rf_acc*100:.2f})")
print(f"ROC_AUC : {rf_roc:.4f}")
```

```python
#Classification report
print(classification_report(y_test,rf_y_pred,target_names=["No Trasanction","Will Transact"]))
print(confusion_matrix(y_test, rf_y_pred))
```

<h2>Gradient Boosting</h2>

```python
from sklearn.ensemble import GradientBoostingClassifier
# Gradient Boosting doesn't have class_weight param directly)
# We use sample_weight instead via compute_sample_weight
from sklearn.utils.class_weight import compute_sample_weight

import warnings
warnings.filterwarnings('ignore')

print("Training Gradient Boosting... (this may take 5-10 minutes on full data)")

gb_model = GradientBoostingClassifier(
    n_estimators=30,     # Number of boosting stages
    learning_rate=0.1,    # How much each tree contributes
    max_depth=1,          # Depth of each tree (shallow = faster, less overfit)
    random_state=42
)




sample_weights = compute_sample_weight(class_weight='balanced', y=y_train)

gb_model.fit(X_train_scaled, y_train, sample_weight=sample_weights)

# Make predictions
gb_pred = gb_model.predict(X_test_scaled)
gb_prob = gb_model.predict_proba(X_test_scaled)[:, 1]

# Evaluate
gb_acc  = accuracy_score(y_test, gb_pred)
gb_roc  = roc_auc_score(y_test, gb_prob)

print(f"\n Gradient Boosting Results:")
print(f"   Accuracy  : {gb_acc:.4f} ({gb_acc*100:.2f}%)")
print(f"   ROC-AUC   : {gb_roc:.4f}")
print("\nDetailed Classification Report:")
print(classification_report(y_test, gb_pred, target_names=['No Transaction', 'Will Transact']))
```

## Challenges Faced and Solutions

### Challenge 1: Imbalanced Dataset

**Problem:**
The dataset contains more customers who will not make transactions (Target = 0) compared to customers who will make transactions (Target = 1).

- No Transaction: 35980 records
- Will Transact: 4020 records

This caused the model to focus more on the majority class and ignore transaction customers.

**Solution:**
- Used `class_weight='balanced'` and custom class weights in Random Forest.
- Used sample weights in Gradient Boosting.
- Evaluated models using ROC-AUC, Recall, and F1-score instead of only accuracy.

---
### Challenge 2: Random Forest Model Bias Towards Majority Class

**Problem:**
Initially, Random Forest predicted most customers as "No Transaction" (Target = 0).

The model achieved high accuracy, but it failed to identify customers who will make transactions (Target = 1).

Example:
- Accuracy was high because the dataset had more Target = 0 records.
- Recall for "Will Transact" class was very low.

**Solution:**
- Applied class balancing techniques.
- Used `class_weight='balanced'` to give more importance to minority class.
- Manually adjusted class weights using:

`class_weight={0:1, 1:5}`

This increased the importance of transaction customers and improved the model's ability to identify potential customers.

- Evaluated the model using Recall, F1-score, and ROC-AUC instead of only accuracy.



### Challenge 3: Model Runtime

**Problem:**
Gradient Boosting took longer training time due to large dataset size.

**Solution:**
- Reduced number of estimators.
- Reduced tree depth.
- Balanced training time and model performance.
